```python
import numpy as np

# Modern approach (recommended): use a Generator
rng = np.random.default_rng()
r = rng.integers(0, 101)  # low=0, high=101 (exclusive), so range is [0, 100]
print(r)

# Alternative: legacy API (still works, but not recommended)
# r = np.random.randint(0, 101)  # [0, 100] inclusive
```

```python
def squares(n):
    for i in range(1,n + 1):
        yield i**2
```

```python
gen = squares(100)
print(gen)
```

```python
for i in gen:
    print(i,end=" ")
```
